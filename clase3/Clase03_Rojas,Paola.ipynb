```python
# IMPORTAMOS LIBRERIAS
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# SUBIMOS EL ARCHIVO
from google.colab import files
files.upload()

# 1. CARGAMOS EL DATASET
df = pd.read_csv("Automobile.csv")

# MOSTRAMOS LAS PRIMERAS FILAS
print(df.head())
```

En esta actividad trabajamos con la librería Matplotlib para visualizar datos del dataset Automobile.csv.

Primero cargamos los datos utilizando Pandas, lo que nos permitió ver las primeras filas y entender la estructura del dataset.

```python
# 2. GRÁFICO DE DISPERSIÓN (mpg vs weight)
plt.figure()
plt.scatter(df["weight"], df["mpg"])
plt.xlabel("Peso del automóvil (weight)")
plt.ylabel("Consumo (mpg)")
plt.title("Relación entre peso y consumo")
plt.show()
```

Luego realizamos un gráfico de dispersión entre el peso (weight) y el consumo (mpg). Este gráfico nos permite observar que, en general, a mayor peso del automóvil, menor es el rendimiento de combustible.

```python
# 3. HISTOGRAMA DE mpg
plt.figure()
plt.hist(df["mpg"], bins=20, color= 'green', edgecolor='black')
plt.xlabel("Consumo (mpg)")
plt.ylabel("Frecuencia")
plt.title("Distribución del consumo de combustible")
plt.show()
```

Después generamos un histograma del mpg para analizar su distribución. Esto nos ayuda a ver cómo se reparten los valores de consumo y detectar si hay valores extremos.
El histograma permite ver cómo se reparten los valores de consumo (mpg). La mayoría de los autos tiene un consumo entre 15 y 30 mpg. También hay algunos casos con valores más altos, mayores a 40 mpg, que pueden corresponder a autos más económicos en combustible.

```python
# 4. BOXPLOT (cylinders y origin)
plt.figure(figsize=(12,6))
df.boxplot(column='cylinders', by= 'origin', grid=True)

plt.title("Boxplot de Cylinders y Origin")
plt.suptitle('')
plt.xlabel('origin')
plt.ylabel('cylinders')
plt.show()
```

También realizamos un boxplot de las variables cylinders y origin. Este tipo de gráfico es útil para identificar posibles valores atípicos (outliers) y ver la dispersión de los datos.
En este ejemplo el diagrama de caja muestra que la cantidad de cilindros cambia según el origen de los autos. En Europa y Japón la mayoría tiene motores de 4 cilindros y no hay mucha diferencia entre ellos, lo que indica que son más eficientes. También se ven algunos valores distintos (atípicos). En cambio, los autos de Estados Unidos tienen más variedad, entre 4 y 8 cilindros, y en general usan motores más grandes.

```python
# 5. GRÁFICO DE BARRAS (model_year y origin)

# Contamos valores
conteo = df.groupby(["model_year", "origin"]).size().unstack()

# Graficamos
conteo.plot(kind="bar")
plt.xlabel("Año del modelo")
plt.ylabel("Cantidad de autos")
plt.title("Distribución por año y origen")
plt.show()
```

Finalmente, creamos un gráfico de barras agrupando por año del modelo y origen. Esto nos permite ver la cantidad de automóviles según estas categorías y comparar su distribución.
En el gráfico se ve que hay más autos de Estados Unidos, sobre todo en los años 70. Europa se mantiene bastante igual, mientras que Japón empieza a crecer al final, lo que podría mostrar un cambio en el mercado.
