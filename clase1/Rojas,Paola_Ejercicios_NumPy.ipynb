Ejercicio 6: Realiza la suma elemento a elemento de los arrays a=[1,2,3], b=[4,5,6].

```python
a = np.array([1,2,3])
b = np.array([4,5,6])
import numpy as np
suma = a + b

print(suma)
```

Ejercicio 7: Multiplica dos matrices 3x3 de enteros aleatorios entre 1 y 10.

```python
import numpy as np
matriz1 = np.random.randint(1, 11, (3, 3))
matriz2 = np.random.randint(1, 11, (3, 3))

resultado = np.dot(matriz1, matriz2)

print("Matriz 1:")
print(matriz1)
print("\nMatriz 2:")
print(matriz2)
print("\nResultado de la multiplicación:")
print(resultado)
```

Ejercicio 8: Crea un array de 15 valores aleatorios entre 1 y 100 e imprime su máximo, mínimo y promedio.

```python
import numpy as np
array_random = np.random.randint(1, 101, 15)

maximo = np.max(array_random)
minimo = np.min(array_random)
promedio = np.mean(array_random)

print("Array generado:", array_random)
print("Máximo:", maximo)
print("Mínimo:", minimo)
print("Promedio:", promedio)
```

Ejercicio 9: Genera una matriz 5x5 con valores del 1 al 25 y extrae su diagonal.

```python
import numpy as np
matriz = np.arange(1, 26).reshape(5, 5)

diagonal = np.diagonal(matriz)

print("Matriz 5x5:")
print(matriz)
print("\nDiagonal de la matriz:")
print(diagonal)
```

Ejercicio 10 : Dada una matriz 5x5 de enteros aleatorios entre 1 y 20, calcula la suma de cada fila y columna

```python
import numpy as np
matriz = np.random.randint(1, 21, (5, 5))

suma_filas = np.sum(matriz, axis=1)
suma_columnas = np.sum(matriz, axis=0)

print("Matriz 5x5:")
print(matriz)
print("\nSuma de cada fila:")
print(suma_filas)
print("\nSuma de cada columna:")
print(suma_columnas)

```
